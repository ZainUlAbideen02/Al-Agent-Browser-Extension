```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_rows = len(df)
declining_rows = len(df[df["trend_direction"] == "down"])
declining_pct = (declining_rows / total_rows) * 100
median_sv_declining = df[df["trend_direction"] == "down"]["search_volume"].median()

print(f"Number 1 — Total Pages Analyzed: {total_rows:,}")
print(f"Number 2 — Declining Pages Rate: {declining_pct:.1f}% ({declining_rows:,} pages)")
print(f"Number 3 — Median Search Volume for Declining Pages: {median_sv_declining}")
```
